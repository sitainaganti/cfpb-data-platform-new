```python
import re

df = (spark.read.option("header", True)
      .option("multiLine", True)
      .option("escape", '"')
      .csv("/Volumes/workspace/default/raw/complaints-2026-11-18_15_25.csv"))

clean_cols = [re.sub(r'[^a-z0-9]+', '_', c.strip().lower()).strip('_') for c in df.columns]
df_clean = df.toDF(*clean_cols)

print(df_clean.columns)   # check the new names
df_clean.write.mode("overwrite").saveAsTable("bronze_complaints")
```
